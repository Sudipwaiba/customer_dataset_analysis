```python
import pandas as pd
```

```python
df1 = pd.read_csv("customer.csv")
df = df1

df['purchased'] = df['purchased'].map({'Yes': 1, 'No': 0})
```

1 How many customers are in this dataset?

```python
df.shape  ## .shape[0] gives only the number of rows 
```

2 Are there any missing values or duplicate rows?

```python
df.isnull().sum()
```

3 What is the youngest age, the oldest age, and the average age?


```python
youngest_age = df['age'].min()
oldest_age = df['age'].max()
avg_age = df['age'].mean()
print(f"The youngest age is : {youngest_age}\n The oldest age is : {oldest_age}\n The average age is : {avg_age}")
```

```python
df.describe()
```

4 How many customers are male and how many are female?


```python
df['gender'].value_counts()
```

5 How many customers are in each education level?

```python
df['education'].value_counts()
```

6 What percentage of customers purchased and what percentage did not?

```python
df['purchased'].value_counts(normalize=True)*100
##  In result O zero denotes No and 1 denotes Yes 
```

7 Do men or women purchase more?

```python
df.groupby('gender')['purchased'].sum()
```

8 Which education level has the most purchases?


```python
df.groupby('education')['purchased'].sum()
```

Based on above results, about 48% of the 50 customers purchased, with ages ranging from 15 to 98 and an average around 54. Females made more purchases than males (14 vs 10), and PG-educated customers purchased the most (10), followed by UG (9) and School (5).
